### Generación del dataset sintético

A partir de la estructura y las distribuciones observadas en el dataset original Hotel Booking Demand, se genera un conjunto de datos sintético de 5,000 reservaciones. Los registros no representan clientes reales y se utilizarán exclusivamente con fines académicos.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 5000
```

```python
datos = pd.DataFrame({
    "id_reserva": [f"RES-{i:05d}" for i in range(1, n + 1)],

    "tipo_hotel": np.random.choice(
        ["City Hotel", "Resort Hotel"],
        size=n,
        p=[0.66, 0.34]
    ),

    "dias_anticipacion": np.clip(
        np.random.gamma(shape=1.3, scale=80, size=n),
        0, 500
    ).astype(int),

    "noches_estancia": np.clip(
        np.random.poisson(lam=3.4, size=n),
        1, 14
    ),

    "numero_huespedes": np.random.choice(
        [1, 2, 3, 4],
        size=n,
        p=[0.22, 0.58, 0.15, 0.05]
    ),

    "segmento_mercado": np.random.choice(
        ["Online TA", "Offline TA/TO", "Direct", "Corporate", "Groups"],
        size=n,
        p=[0.47, 0.20, 0.12, 0.11, 0.10]
    ),

    "tipo_cliente": np.random.choice(
        ["Transient", "Transient-Party", "Contract", "Group"],
        size=n,
        p=[0.75, 0.20, 0.04, 0.01]
    ),

    "tipo_deposito": np.random.choice(
        ["No Deposit", "Non Refund", "Refundable"],
        size=n,
        p=[0.87, 0.12, 0.01]
    ),

    "tarifa_diaria": np.clip(
        np.random.normal(loc=102, scale=35, size=n),
        35, 300
    ).round(2),

    "solicitudes_especiales": np.random.choice(
        [0, 1, 2, 3, 4],
        size=n,
        p=[0.55, 0.28, 0.12, 0.04, 0.01]
    ),

    "cancelaciones_previas": np.random.choice(
        [0, 1, 2, 3],
        size=n,
        p=[0.91, 0.06, 0.02, 0.01]
    )
})

datos.head(10)
```

```python
#construyo una probabilidad de cancelación relacionada con algunas características, añadiendo aleatoriedad suficiente para que el resultado no sea perfectamente predecible.
# La intención es obtener una tasa cercana al 37 % observado en el dataset original, sin copiar sus registros.
```

```python
# Puntaje base de cancelación. Un valor menos negativo, generaría más cancelaciones. Uno más negativo, genera menos.
puntaje = np.full(n, -1.20)

# Mayor anticipación incrementa el riesgo
puntaje += datos["dias_anticipacion"] * 0.004

# Historial de cancelaciones incrementa el riesgo
puntaje += datos["cancelaciones_previas"] * 0.75

# Efecto del segmento de mercado
puntaje += np.where(
    datos["segmento_mercado"] == "Online TA",
    0.35,
    0
)

puntaje += np.where(
    datos["segmento_mercado"] == "Groups",
    0.20,
    0
)

# Efecto del tipo de depósito
puntaje += np.where(
    datos["tipo_deposito"] == "Non Refund",
    0.65,
    0
)

puntaje -= np.where(
    datos["tipo_deposito"] == "Refundable",
    0.25,
    0
)

# Los clientes recurrentes o de contrato presentan menor riesgo
puntaje -= np.where(
    datos["tipo_cliente"] == "Contract",
    0.35,
    0
)

# Las solicitudes especiales indican mayor interacción con el hotel
puntaje -= datos["solicitudes_especiales"] * 0.28

# Las estancias más largas presentan un ligero incremento del riesgo
puntaje += np.where(
    datos["noches_estancia"] >= 7,
    0.20,
    0
)

# Agregar variación aleatoria
puntaje += np.random.normal(
    loc=0,
    scale=0.55,
    size=n
)

# Convertir el puntaje a una probabilidad entre 0 y 1
probabilidad_cancelacion = 1 / (1 + np.exp(-puntaje))

probabilidad_cancelacion[:10]
```

```python
#Se evalúa la probabilidad correspondiente a cada reservación. 1: reservación cancelada, 0: reservación no cancelada. Dos reservaciones con características similares todavía pueden tener resultados diferentes. Esto evita que el dataset sea totalmente determinista

datos["cancelada"] = np.random.binomial(
    n=1,
    p=probabilidad_cancelacion,
    size=n
)

datos.head()
```

```python
#Validar la tasa de cancelación

conteo_cancelaciones = datos["cancelada"].value_counts().sort_index()

proporcion_cancelaciones = (
    datos["cancelada"]
    .value_counts(normalize=True)
    .sort_index()
    .round(3)
)

print("Conteo:")
print(conteo_cancelaciones)

print("\nProporción:")
print(proporcion_cancelaciones)

print(
    "\nTasa de cancelación:",
    f"{datos['cancelada'].mean():.1%}"
)

#Primero se obtuvo un porcentaje de cancelación del 32% y se ajustó el puntaje base de cancelación a -1.20 para acercarse al 37% observado en el dataset original.
```

```python
#Se revisa que las relaciones tengan sentido para comprobar que las relaciones programadas aparecen en los datos.

resumen_segmentos = (
    datos.groupby("segmento_mercado")["cancelada"]
    .agg(
        reservaciones="count",
        tasa_cancelacion="mean"
    )
    .sort_values("tasa_cancelacion", ascending=False)
)

resumen_segmentos["tasa_cancelacion"] = (
    resumen_segmentos["tasa_cancelacion"] * 100
).round(1)

resumen_segmentos
```

```python
resumen_depositos = (
    datos.groupby("tipo_deposito")["cancelada"]
    .agg(
        reservaciones="count",
        tasa_cancelacion="mean"
    )
    .sort_values("tasa_cancelacion", ascending=False)
)

resumen_depositos["tasa_cancelacion"] = (
    resumen_depositos["tasa_cancelacion"] * 100
).round(1)

resumen_depositos
```

```python
tasa_global = datos["cancelada"].mean()

print(f"Tasa global de cancelación: {tasa_global:.1%}")
print("\nDistribución:")
print(datos["cancelada"].value_counts())
```

```python
datos_app = datos.copy()
```

```python
# Se generan 25 valores faltantes en tarifa_diaria: 0.5 % y 15 valores faltantes en tipo_cliente: 0.3 %

np.random.seed(42)

indices_tarifa = np.random.choice(
    datos_app.index,
    size=25,
    replace=False
)

indices_cliente = np.random.choice(
    datos_app.index.difference(indices_tarifa),
    size=15,
    replace=False
)

datos_app.loc[indices_tarifa, "tarifa_diaria"] = np.nan
datos_app.loc[indices_cliente, "tipo_cliente"] = np.nan
```

```python
#Generar duplicados

duplicados = datos_app.sample(
    n=8,
    random_state=42
).copy()

datos_app = pd.concat(
    [datos_app, duplicados],
    ignore_index=True
)

#El dataset final tiene ahora 5008  registros
```

```python
print("Dimensiones finales:", datos_app.shape)

print("\nValores nulos:")
print(datos_app.isnull().sum())

print("\nFilas duplicadas:")
print(datos_app.duplicated().sum())
```

```python
ruta_salida = "data/reservas_hotel_sinteticas.csv"

datos_app.to_csv(
    ruta_salida,
    index=False
)

print(f"Dataset guardado correctamente en: {ruta_salida}")
```
